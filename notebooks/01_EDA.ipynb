# Part 2: Exploratory Data Analysis (EDA)

In this notebook, we will explore the **PlantVillage** dataset that was downloaded.
We will check:
1. The number of classes (diseases/crops)
2. The distribution of images per class (to check for class imbalance)
3. Visualize some sample images to understand what our model will learn.

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from collections import Counter
import random

# Define dataset paths
dataset_dir = '../dataset/archive (2)/PlantVillage'
train_dir = os.path.join(dataset_dir, 'train')
val_dir = os.path.join(dataset_dir, 'val')
```

```python
# 1. List all classes
classes = os.listdir(train_dir)
print(f"Total number of classes: {len(classes)}")
print("Sample classes:", classes[:5])
```

```python
# 2. Check class distribution (Number of images per class)
class_counts = {}
for class_name in classes:
    class_path = os.path.join(train_dir, class_name)
    num_images = len(os.listdir(class_path))
    class_counts[class_name] = num_images

# Sort and plot the distribution
sorted_counts = dict(sorted(class_counts.items(), key=lambda item: item[1], reverse=True))

plt.figure(figsize=(15, 6))
plt.bar(sorted_counts.keys(), sorted_counts.values())
plt.xticks(rotation=90)
plt.title("Number of images per class in Training Set")
plt.ylabel("Number of Images")
plt.xlabel("Class")
plt.show()
```

```python
# 3. Visualize a few random images
num_samples = 4
random_classes = random.sample(classes, num_samples)

fig, axes = plt.subplots(1, num_samples, figsize=(15, 5))

for i, class_name in enumerate(random_classes):
    class_path = os.path.join(train_dir, class_name)
    image_file = random.choice(os.listdir(class_path))
    img_path = os.path.join(class_path, image_file)
    
    img = mpimg.imread(img_path)
    axes[i].imshow(img)
    axes[i].set_title(class_name.split('___')[-1])  # Show just the disease name for brevity
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
